```python
from cluster_intrep_repo.utils import DOMAIN_PHRASES
```

```python
mapping = {
    "actions": {
        "attack": "pick up",
        "succumb": "put down",
        "overcome": "stack",
        "feast": "unstack"
    },
    "predicates": {
        "planet": "on table",
        "province": "clear",
        "harmony": "handempty",
        "craves": "on",
        "pain": "holding"
    }
}
```

```python
phrases = {
    k: {
        "actions": {
            mapping["actions"][kk]: vv for kk, vv in v["actions"].items()
        },
        "predicates": {
            mapping["predicates"][kk]: vv for kk, vv in v["predicates"].items()
        }
    } for k, v in DOMAIN_PHRASES.items()
}
```

```python
for k, v in phrases.items():
    print(k)
    print("Actions:")
    for kk, vv in v["actions"].items():
        print(f"\t{kk}: {vv}")
    print("Predicates:")
    for kk, vv in v["predicates"].items():
        print(f"\t{kk}: {vv}")
    print("\n")

```

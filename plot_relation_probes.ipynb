```python
import os
import json
os.chdir("..")
```

```python
with open("nohup.out") as f:
    logs = f.readlines()
```

```python
from collections import defaultdict

pattern =  "Top block: {top}, Bottom block: {botom}, Hits:"

blocks = [chr(65 + i) for i in range(6)]

def parse_line(line, line_start):
    line = line[len(line_start):]
    line = line.strip()

    acc, f1 = line.split(", F1: ")

    acc = float(acc)
    f1 = float(f1)

    return acc, f1

stats = defaultdict(list)

for top_block in blocks:
    for bottom_block in blocks:
        if top_block == bottom_block:
            continue
        line_start = pattern.format(top=top_block, botom=bottom_block)
        for line in logs:
            if line.startswith(line_start):
                stats[(top_block, bottom_block)].append(parse_line(line, line_start))
```

```python
stats
```

```python
outputs = []

for k, v in stats.items():
    outputs.append({
        "top": k[0],
        "bottom": k[1],
        "combined f1": v[-3][1],
        "correct f1": v[-2][1],
        "incorrect f1": v[-1][1],
    })

with open("6-blocks-pairwise-relations.json", "w") as f:
    json.dump(outputs, f, indent=4)
```

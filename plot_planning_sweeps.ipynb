```python
import json
import pandas as pd

import plotly.express as px
```

```python
import os

os.chdir("..")
```

```python
from pathlib import Path

def process_run(run_dir: Path):
    df = None
    params = None

    for fname in run_dir.rglob("*.csv"):
        df = pd.read_csv(fname)

    for fname in run_dir.rglob("**/params.json"):
        with open(fname) as f:
            params = json.load(f)

    return df["val_acc"].max(), df["train_loss"].min(), params["train_loop_config"]
```

```python
target_dir = Path("ray_results")

results = []

for run_dir in target_dir.glob("TorchTrainer_2025-03-16_1*"):

    try:
        f1, loss, params = process_run(run_dir)
        probe_type = params["probe_type"]
        n_shift_tokens = params["n_shift_tokens"]
        n_prev_tokens = params["n_prev_tokens"]
        tgt_action = params.get("tgt_action", 0)

        results.append({
            "f1": f1,
            "loss": loss,
            "probe type": probe_type,
            "n_shift_tokens": abs(n_shift_tokens),
            # "n_prev_tokens": n_prev_tokens,
            "tgt_action": tgt_action
        })
    except:
        pass
```

```python
df = pd.DataFrame.from_records(results)

df
```

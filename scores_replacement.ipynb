```python
files = [
    f"results_node_{i}.json" for i in range(2, 5)
]
```

```python
from collections import defaultdict

exp_names = []

for file in files:
    with open(file, "r") as f:
        data = json.load(f)
    for x in data["1"]:
        exp_names.append(x)
```

```python
template = 'qwq-32b-replaced-st-a-full-end-1000-2500-{scale}-l-{layer}-fix-rescale{extra}'
```

```python
results = {x: {} for x in exp_names}

for file in files:
    with open(file, "r") as f:
        data = json.load(f)
    for mn in data:
        for k, v in data[mn].items():
            results[k][mn] = v
```
